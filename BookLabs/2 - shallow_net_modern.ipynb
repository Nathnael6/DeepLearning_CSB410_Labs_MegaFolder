# Shallow Neural Network in Keras: Network B (modern defaults)

Same data and same three-layer structure as `2_3_4-shallow_net_in_keras.ipynb`, with the settings most current tutorials use: 128 ReLU neurons, cross-entropy loss, the Adam optimizer, and 5 epochs.

This is the code shown as **Network B** in the two-network version of the Week 2 slides.

#### Load dependencies

```python
import numpy as np
import matplotlib.pyplot as plt
import keras
from keras import layers
```

#### Load data

```python
(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data()
```

```python
X_train.shape, y_train.shape, X_test.shape, y_test.shape
```

#### Preprocess data

Scale pixel values from 0–255 down to 0–1. The images stay 28 × 28 and the labels stay as integers 0–9; the model handles the flattening.

```python
X_train = X_train / 255.0
X_test = X_test / 255.0
```

#### Design neural network architecture

```python
model = keras.Sequential([
    keras.Input(shape=(28, 28)),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax')
])
```

```python
model.summary()
```

```python
(784*128)+128
```

```python
(128*10)+10
```

#### Configure model

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

#### Train!

Re-running this cell continues training from the current weights. To start over, re-run the architecture cell first.

```python
model.fit(X_train, y_train,
          epochs=5,
          validation_data=(X_test, y_test))
```

#### Evaluate

```python
loss, accuracy = model.evaluate(X_test, y_test)
print("Test accuracy:", accuracy)
```

#### Make predictions

```python
predictions = model.predict(X_test)
predicted_labels = np.argmax(predictions, axis=1)
```

```python
predictions[0]
```

```python
predicted_labels[0], y_test[0]
```

#### Visualize a prediction

```python
plt.imshow(X_test[0], cmap='gray')
plt.title("Predicted Label: " + str(predicted_labels[0]))
plt.show()
```

#### Find the ones it got wrong

```python
wrong = np.where(predicted_labels != y_test)[0]
print(len(wrong), "mistakes out of", len(y_test))
```

```python
plt.imshow(X_test[wrong[0]], cmap='gray')
plt.title(f"Predicted {predicted_labels[wrong[0]]}, actually {y_test[wrong[0]]}")
plt.show()
```
